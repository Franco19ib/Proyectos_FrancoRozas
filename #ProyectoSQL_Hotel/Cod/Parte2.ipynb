```sql
SELECT
    COUNT(*) AS reservas_canceladas_ta_to,
    ROUND(COUNT(*) * 1.5, 2) AS perdida_marketing_usd
FROM workspace.hotel_db.tb_reservas
WHERE distribution_channel = 'TA/TO'
  AND is_canceled = 1;
```

```sql
SELECT
    COUNT(*) AS cancelaciones_tardias,
    COUNT(*) * 120 AS perdida_operativa_usd
FROM workspace.hotel_db.tb_reservas
WHERE is_canceled = 1
  AND DATE_DIFF(arrival_date, reservation_status_date) BETWEEN 0 AND 2;
```

```sql
SELECT r.is_canceled, r.country AS cod_pais, p.nombre_pais
FROM workspace.hotel_db.tb_reservas AS r
LEFT JOIN workspace.hotel_db.tb_paises AS p
    ON r.country = p.cod_pais;
```

```sql
SELECT p.nombre_pais, COUNT(*) AS reservas_canceladas
FROM workspace.hotel_db.tb_reservas AS r
INNER JOIN workspace.hotel_db.tb_paises AS p
    ON r.country = p.cod_pais
WHERE r.is_canceled = 1
GROUP BY p.nombre_pais
ORDER BY reservas_canceladas DESC
LIMIT 5;
```

```sql
SELECT
  r.distribution_channel,
  v.nombre_vendedor
FROM workspace.hotel_db.tb_reservas r
FULL OUTER JOIN workspace.hotel_db.tb_vendedores v
  ON r.id_vendedor = v.id_vendedor;
```

```sql
SELECT f.id_reserva, f.monto_facturado, r.hotel
FROM workspace.hotel_db.tb_reservas AS r
RIGHT JOIN workspace.hotel_db.tb_facturas AS f
    ON r.id_reserva = f.id_reserva;
```
