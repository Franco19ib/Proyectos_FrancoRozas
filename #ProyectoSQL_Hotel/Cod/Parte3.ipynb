```sql
SELECT
  CASE
    WHEN TRY_CAST(children AS INT) > 0 OR TRY_CAST(babies AS INT) > 0 THEN 'Con Hijos'
    ELSE 'Sin Hijos'
  END AS segmento,
  COUNT(*)                          AS total_reservas,
  AVG(TRY_CAST(is_canceled AS INT)) AS tasa_cancelacion
FROM workspace.hotel_db.tb_reservas
GROUP BY
  CASE
    WHEN TRY_CAST(children AS INT) > 0 OR TRY_CAST(babies AS INT) > 0 THEN 'Con Hijos'
    ELSE 'Sin Hijos'
  END
ORDER BY tasa_cancelacion DESC;
```

```sql
SELECT
  total_of_special_requests,
  COUNT(*)         AS total_reservas,
  AVG(is_canceled) AS tasa_cancelacion
FROM workspace.hotel_db.tb_reservas
GROUP BY total_of_special_requests
ORDER BY total_of_special_requests ASC;
```

```sql
SELECT
  CASE
    WHEN booking_changes = 0 THEN 'Sin Cambios'
    WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
    ELSE 'Más de 3 cambios'
  END AS grupo_cambios,
  COUNT(*)         AS total_reservas,
  AVG(is_canceled) AS tasa_cancelacion
FROM workspace.hotel_db.tb_reservas
GROUP BY
  CASE
    WHEN booking_changes = 0 THEN 'Sin Cambios'
    WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
    ELSE 'Más de 3 cambios'
  END
ORDER BY tasa_cancelacion ASC;
```

```sql
SELECT
  hotel,
  COUNT(*) AS total_cancelaciones,
  AVG(adr) AS adr_promedio_canceladas
FROM workspace.hotel_db.tb_reservas
WHERE is_canceled = 1
GROUP BY hotel
ORDER BY adr_promedio_canceladas DESC;
```
