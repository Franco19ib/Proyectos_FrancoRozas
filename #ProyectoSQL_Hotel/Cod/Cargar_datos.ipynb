```python
import os
csv_file = "hotel_bookings.csv"

file_path = f"file:{folder_path}/{csv_file}"

# 2. LEER EL CSV CON SPARK
df_spark = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .option("sep", ",")
    .option("encoding", "UTF-8")
    .option("mode", "PERMISSIVE")
    .load(file_path)
)

# 3. REVISAR LOS DATOS
display(df_spark.limit(10))

# Ver los tipos de datos de las columnas
df_spark.printSchema()

# Crear esquema
spark.sql("""
CREATE SCHEMA IF NOT EXISTS workspace.hotel_db
""")

# Nombre de la tabla
table_name = "workspace.hotel_db.tb_reservas"

# Guardar CSV como tabla Delta
df_spark.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(table_name)

print("Carga completada")
print("Filas:", df_spark.count())
print("Columnas:", len(df_spark.columns))
```
